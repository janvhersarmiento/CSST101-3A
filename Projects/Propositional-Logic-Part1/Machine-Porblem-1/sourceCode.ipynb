# Enhanced Mini Expert System: University Logic Rules
# With CSV Logging for Record-Keeping

```python
import csv
from datetime import datetime
```

# Logic Functions

```python
def impl(P,Q):
    return (not P) or Q

def tf(b: bool) -> str:
    return 'T' if b else 'F'
```

# Logger

```python
def log_result(student_name, rule_name, result):
    with open("logic_results.csv", 'a', newline="") as file:
        writer = csv.writer(file)
        writer.writerow([datetime.now().strftime("%Y-%m-%d %H:%M:%S"),student_name,rule_name,result])
```

# Rule 1: Attendance

```python
def attendance_rule(student_name):
    print("\n--- Attendance Rule Checker ---")
    
    # Input phase
    late_input = input("Is the student late? (T/F): ").strip().upper()
    excuse_input = "F"

    if late_input == "T":
        excuse_input = input("Did the student bring an excuse letter? (T/F): ").strip().upper()

    P = (late_input == "T")
    Q = (excuse_input == "T")

    result = impl(P, Q)
    outcome = "Satisfied ✅" if result else "Violated ❌"

    print(f"P = {tf(P)} (Late), Q = {tf(Q)} (Brought Excuse Letter)")
    print("Logical Rule: If Late → Must Have Excuse Letter")
    print("Result:", outcome)

    log_result(student_name, "Attendance Rule", outcome)
```

# Rule 2: Grading

```python
def grading_rule(student_name):
    print("\n--- Grading Rule Checker ---")
    try:
        grade = float(input("Enter Student Grade: ").strip())
    except ValueError:
        print("Invalid Grade Input! Please enter a number.")
        return

    P = grade >= 75
    Q = P

    result = impl(P, Q)
    outcome = "Satisfied ✅" if result else "Violated ❌"

    if grade >= 75:
        status = "Passed ✅"
    else:
        status = "Failed ❌"
        outcome = "Violated ❌"

    print(f"P = {tf(P)} (Grade ≥ 75), Q = {tf(Q)} (Student Passes)")
    print("Logical Rule: If Grade ≥ 75 → Student Passes")
    print(f"Grade: {grade} — Status: {status}")
    print("Result:", outcome)

    log_result(student_name, "Grading Rule", status)
```

# Rule 3: Login System

```python
def login_rule(student_name):
    print("\n--- Login Rule Checker ---")
    correct_password = "admin123"
    attempt = input("Enter Password: ")

    P = (attempt == correct_password)
    Q = (attempt == correct_password)

    result = impl(P, Q)
    outcome = "Satisfied ✅" if result else "Violated ❌"

    print(f"P = {tf(P)} (Password Correct), Q = {tf(Q)} (Access Granted)")
    print("Result:", outcome)

    log_result(student_name, "Login Rule", outcome)
```

# Rule 4: Academic Status

```python
def academic_status(student_name):
    print("\n--- Academic Status Rule Checker ---")
    try:
        gpa = float(input(f"Enter {student_name}'s GPA: ").strip())
    except ValueError:
        print("Invalid input! Please enter a numeric GPA.")
        return

    P = (gpa <= 3.00)
    Q = (gpa <= 2.50)

    result = impl(P, Q)
    outcome = "Satisfied ✅" if result else "Violated ❌"

    if gpa <= 1.75:
        status = "Honors Status"
    elif gpa <= 2.50:
        status = "Regular Status"
    elif gpa <= 3.00:
        status = "Probationary Status"
    else:
        status = "At Risk of Dismissal"

    print(f"P = {tf(P)} (GPA within Passing Range), Q = {tf(Q)} (Good Status)")
    print(f"Determined Status: {status}")
    print("Result:", outcome)

    log_result(student_name, "Academic Status Rule", f"{outcome} - {status}")
```

# Rule 5: Bonus Points

```python
def bonus_rule(student_name):
    print("\n--- Bonus Points Eligibility Checker ---")
    
    regular_input = input("Does the student have regular attendance? (T/F): ").strip().upper()
    regular = (regular_input == "T")

    P = regular
    Q = regular

    result = impl(P, Q)
    outcome = "Satisfied ✅" if result else "Violated ❌"

    if not regular:
        Q = False
        result = impl(P, Q)
        outcome = "Violated ❌"

    print(f"P = {tf(P)} (Regular Attendance), Q = {tf(Q)} (Bonus Eligible)")
    print("Logical Rule: If Regular Attendance → Eligible for Bonus")
    print("Result:", outcome)

    log_result(student_name, "Bonus Rule", outcome)
```

## MAIN

```python
def main():
    print("=== University Logic Rule System ===")
    student_name = input("Enter Student Name: ".strip())

    while True:
        print("\n==============================")
        print("    Main Menu")
        print("==============================")

        print("1) Attendance Rule Checker")
        print("2) Grading Rule Checker")
        print("3) Login System Rule Checker")
        print("4) Academic Status Rule Checker")
        print("5) Bonus Points Checker")
        print("6) Exit")

        choice = input("Choose an Option: ").strip()

        match choice:
            case "1": attendance_rule(student_name)
            case "2": grading_rule(student_name)
            case "3": login_rule(student_name)
            case "4": academic_status(student_name)
            case "5": bonus_rule(student_name)
            case "6":
                print("Exiting... Results saved to logic_results.csv")
                break
            case _: print("Unknown Choice")


if __name__ == "__main__":
    # Create CSV with headers if not exist
    with open("logic_results.csv", "a", newline="") as file:
        writer = csv.writer(file)
        if file.tell() == 0: # Only write the header if file is empty
            writer.writerow(["Timestamp", "Student Name", "Rule", "Result"])

main()
```
