Translate the following statements into predicate logic and show the reasoning steps:<br>
<br>1. “All humans are mortal.” “Socrates is a human.” → Prove that Socrates is mortal.<br>
**Translation to Predicate Logic:**

   - Let Human(x) mean "x is a human."
   - Let Mortal(x) mean "x is mortal."
   - Statement 1: ∀x (Human(x) → Mortal(x))
   - Statement 2: Human(Socrates)
   - To Prove: Mortal(Socrates)

**Reasoning Steps:**

   1. Start with the universal rule:
      ∀x (Human(x) → Mortal(x))
   2. Apply Universal Instantiation: We can substitute the variable x with the constant Socrates.
      Human(Socrates) → Mortal(Socrates)
   3. Use the known fact:
      Human(Socrates)
   4. Apply Modus Ponens: From the statements in Step 2 (P → Q) and Step 3 (P), we can infer Q.
      Mortal(Socrates)

**Conclusion:** We have successfully proven that Socrates is mortal.<br>

      Teacher(Mark) → ∃y (Student(y) ∧ Advises(Mark, y))
   3. Use the known fact:
      Teacher(Mark)
   4. Apply Modus Ponens: From the statements in Step 2 (P → Q) and Step 3 (P), we can infer Q.
      ∃y (Student(y) ∧ Advises(Mark, y))
